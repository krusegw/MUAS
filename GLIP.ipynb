```python
from gilp.simplex import LP
```

```python
from gilp.visualize import simplex_visual
```

```python
import numpy as np
```

```python
A = np.array([[33,58],
              [100,-75],
              [-1, 0],
              [0, -1]])
b = np.array([[60000],
              [150000],
              [0],
              [0]])
c = np.array([[1],
              [1]])

lp = LP(A,b,c)
```

```python
simplex_visual(lp).show()
```
