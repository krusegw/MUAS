```python
from gilp.simplex import LP
```

```python
from gilp.visualize import simplex_visual
```

```python
import numpy as np
```

```python
A = np.array([[33,58],
              [100,-75],
              [-1, 0],
              [0, -1]])
b = np.array([[60000],
              [150000],
              [0],
              [0]])
c = np.array([[1],
              [1]])
lp = LP(A,b,c)
print(A)
print(b)
print(c)
```

```python
A = np.array([[2,2],
              [5,3],
              [-1, 0],
              [0, -1]])
b = np.array([[8],
              [15],
              [0],
              [0]])
c = np.array([[120],
              [100]])
lp = LP(A,b,c)
print(A)
print(b)
print(c)
```

```python
simplex_visual(lp).show()
```
